##  Model Comparison & Evaluation Metrics

---


This section aggregates and compares the evaluation metrics (Accuracy, Precision, Recall, F1-Score) for all 6 Machine Learning and Deep Learning models used in this project.

```python

# 1. MOUNT GOOGLE DRIVE

from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define Google Drive save directory
save_dir = '/content/drive/MyDrive/Spam_Project_Results'
os.makedirs(save_dir, exist_ok=True)
print(f"Results will be saved to: {save_dir}")


# 2. AGGREGATE RESULTS FROM ALL TEAM MEMBERS


data = {
    'Model Type': [
        'Machine Learning', 'Machine Learning', 'Machine Learning',
        'Deep Learning', 'Deep Learning', 'Deep Learning'
    ],
    'Model Name': [
        'Logistic Regression', 'SVM', 'Random Forest',
        'LSTM', 'GRU', 'BERT'
    ],
    'Accuracy': [0.9577, 0.9776, 0.9677, 0.9581, 0.8752, 0.9826],
    'Precision': [0.9842, 0.9800, 0.9746, 0.9623, 0.7165, 0.9776],
    'Recall': [0.8736, 0.9400, 0.9103, 0.8961, 0.9691, 0.9611],
    'F1-Score': [0.9256, 0.9600, 0.9414, 0.9280, 0.8239, 0.9693]
}

# Create DataFrame
results_df = pd.DataFrame(data)

# Display the aggregated comparison table
print("=================== ALL MODELS PERFORMANCE COMPARISON ===================")
display(results_df)

# Save Comparison Table to Google Drive as CSV
csv_path = os.path.join(save_dir, 'model_comparison_results.csv')
results_df.to_csv(csv_path, index=False)
print(f"Comparison table successfully saved to Drive: {csv_path}")


# 3. VISUALIZE AND SAVE MODEL COMPARISON CHART


# Set plot style and dimensions
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Reshape DataFrame for Seaborn visualization
df_melted = results_df.melt(
    id_vars=["Model Type", "Model Name"],
    var_name="Metric",
    value_name="Score"
)

# Generate grouped bar chart
ax = sns.barplot(x="Model Name", y="Score", hue="Metric", data=df_melted, palette="Set2")

# Chart Customizations
plt.title("Performance Comparison: Machine Learning vs Deep Learning Models", fontsize=14, fontweight='bold')
plt.xlabel("Model Name", fontsize=12)
plt.ylabel("Score (0.0 - 1.0)", fontsize=12)
plt.ylim(0.8, 1.05)  # Adjust lower/upper limits based on your actual score ranges
plt.legend(loc='lower right', frameon=True)
plt.xticks(rotation=15)

# Display exact scores on top of each bar
for p in ax.patches:
    if p.get_height() > 0:
        ax.annotate(f"{p.get_height():.3f}",
                    (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='center',
                    xytext=(0, 5),
                    textcoords='offset points',
                    fontsize=8,
                    rotation=90)

plt.tight_layout()

# Save Chart Image to Google Drive
chart_path = os.path.join(save_dir, 'model_comparison_chart.png')
plt.savefig(chart_path, dpi=300, bbox_inches='tight')
print(f"Comparison chart successfully saved to Drive: {chart_path}")

plt.show()
```
